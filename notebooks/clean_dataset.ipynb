```python
import pandas as pd

def main():
    df = pd.read_csv(
        "../data/insurance_data.csv",
        on_bad_lines="skip",
        low_memory=False
    )

    df_clean = df.dropna()

    df_clean.to_csv("../data/insurance_data_clean.csv", index=False)

if __name__ == "__main__":
    main()
```

## Data Quality Handling

During ingestion, the dataset contained malformed rows with inconsistent field counts. These were handled using `on_bad_lines='skip'` to ensure successful parsing while preserving overall dataset integrity.

A cleaned dataset was then generated by removing missing values.
